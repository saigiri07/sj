```python
# Create an empty subplot with 2 rows and 2 columns (4 subplots total)
import matplotlib.pyplot as plt

# Create figure and 2x2 axes array
fig, axes = plt.subplots(2, 2)

# Display the empty subplots
plt.show()
```

Notice how the subplot has multiple figures. Now let's add data to each axes.

```python
# Create the same plot as above with 2 rows and 2 columns and figsize of (10, 5)
import numpy as np
import matplotlib.pyplot as plt

# Prepare data
X = np.linspace(0, 10, 100)
Y = X / 2

random_x = np.random.rand(10)
random_y = np.random.rand(10)

favorite_foods = {
    "Pizza": 12.99,
    "Sushi": 18.50,
    "Ice Cream": 4.75
}
foods = list(favorite_foods.keys())
prices = list(favorite_foods.values())

normal_data = np.random.randn(1000)

# Create figure and axes with specified figsize
fig, axes = plt.subplots(2, 2, figsize=(10, 5))

# Top-left: Line plot X vs X/2
axes[0, 0].plot(X, Y)
axes[0, 0].set_title("X vs X/2")
axes[0, 0].set_xlabel("X")
axes[0, 0].set_ylabel("X/2")

# Top-right: Scatter plot of 10 random points
axes[0, 1].scatter(random_x, random_y)
axes[0, 1].set_title("Scatter Plot of Random Points")
axes[0, 1].set_xlabel("Random X")
axes[0, 1].set_ylabel("Random Y")

# Bottom-left: Bar graph of favorite foods
axes[1, 0].bar(foods, prices)
axes[1, 0].set_title("Favorite Food Prices")
axes[1, 0].set_xlabel("Food")
axes[1, 0].set_ylabel("Price ($)")

# Bottom-right: Histogram of normal distribution
axes[1, 1].hist(normal_data, bins=30, edgecolor='black')
axes[1, 1].set_title("Histogram of Normally Distributed Data")
axes[1, 1].set_xlabel("Value")
axes[1, 1].set_ylabel("Frequency")

# Adjust layout to prevent overlap
plt.tight_layout()

# Show plot
plt.show()
```

```python
import pandas as pd
# Load CSV into DataFrame
car_sales = pd.read_csv('/content/carsales.csv')
print(car_sales.head())

# View the first 5 rows
print(car_sales.head())
```

```python
# Try to plot the 'Price' column using the plot() function
import pandas as pd
import matplotlib.pyplot as plt

# Clean the 'Price' column:
# 1. Remove '$' and ',' symbols
# 2. Remove last 3 characters (".00")
# 3. Convert to numeric

car_sales['Price_clean'] = (
    car_sales['Price']
    .str.replace('[\$,]', '', regex=True)  # Remove $ and ,
    .str.replace('\.00$', '', regex=True)  # Remove trailing ".00"
    .astype(float)                         # Convert to float
)

# Plot the cleaned Price column
car_sales['Price_clean'].plot()

# Customize the plot
plt.xlabel('Index')
plt.ylabel('Price ($)')
plt.title('Car Sales Price Plot')

plt.show()

```

Why doesn't it work?

Hint: It's not numeric data.

In the process of turning it to numeric data, let's create another column which adds the total amount of sales and another one which shows what date the car was sold.

Hint: To add a column up cumulatively, look up the cumsum() function. And to create a column of dates, look up the date_range() function.

```python
# Remove the symbols, the final two numbers from the 'Price' column and convert it to numbers
# Example: Price values like '$12,345.00' -> remove $ and commas, drop last two chars, convert to number

# Remove symbols like $ and commas
car_sales['Price_clean'] = car_sales['Price'].str.replace('[\$,]', '', regex=True)

# Remove last two characters (e.g., ".00" or any last two chars)
car_sales['Price_clean'] = car_sales['Price_clean'].str[:-2]

# Convert to numeric (float or int)
car_sales['Price_clean'] = pd.to_numeric(car_sales['Price_clean'])

print(car_sales[['Price', 'Price_clean']].head())
```

```python
# 1. Clean 'Price' column first (as done previously)
car_sales['Price_clean'] = car_sales['Price'].str.replace('[\$,]', '', regex=True).str[:-2]
car_sales['Price_clean'] = pd.to_numeric(car_sales['Price_clean'])

# 2. Add 'Total Sales' column (cumulative sum of cleaned prices)
car_sales['Total Sales'] = car_sales['Price_clean'].cumsum()

# 3. Add 'Sale Date' column with successive dates starting from today
today = pd.to_datetime('today').normalize()
car_sales['Sale Date'] = pd.date_range(start=today, periods=len(car_sales), freq='D')

# 4. View the updated DataFrame
print(car_sales.head())
```

Now we've got a numeric column (`Total Sales`) and a dates column (`Sale Date`), let's visualize them.

```python
# Use the plot() function to plot the 'Sale Date' column versus the 'Total Sales' column
# Plot 'Sale Date' vs 'Total Sales'
car_sales.plot(x='Sale Date', y='Total Sales')

# Add labels and title
plt.xlabel('Sale Date')
plt.ylabel('Total Sales ($)')
plt.title('Total Car Sales Over Time')

# Show the plot
plt.show()
```

```python
# Replicate the above plot in whichever way you see fit
import matplotlib.pyplot as plt

# 1. Create DataFrame with patients over 50
over_50 = heart_disease[heart_disease['age'] > 50]

# 2. Create histogram subplots of numeric columns for patients over 50
axes = over_50.plot.hist(
    subplots=True,
    figsize=(10, 30),
    bins=30,
    edgecolor='black',
    alpha=0.7  # Slight transparency
)

# 3. Customize each subplot by adding a vertical mean line
for ax, col in zip(axes, over_50.select_dtypes(include='number').columns):
    mean_val = over_50[col].mean()
    ax.axvline(mean_val, color='red', linestyle='--', linewidth=1.5)
    ax.set_title(f'{col} Distribution (Mean line in red)')
    ax.grid(True)

plt.tight_layout()
plt.show()

```

Wonderful, you've changed the style of the plots and the figure is looking different but the dots aren't a very good colour.

Let's change the `cmap` parameter of `scatter()` as well as the `color` parameter of `axhline()` to fix it.

Completing this step correctly should result in a figure which looks like plot3

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data (replace with your actual data if needed)
x = np.random.rand(100)
y = np.random.rand(100)

fig, ax = plt.subplots()

# Scatter plot with 'winter' colormap
scatter = ax.scatter(x, y, c=y, cmap='winter')

# Add colorbar for the colormap
fig.colorbar(scatter, ax=ax)

# Add a horizontal mean line (mean of y)
mean_y = np.mean(y)
ax.axhline(mean_y, color='red', linestyle='--', linewidth=1.5)

# Customize plot
ax.set_title('Scatter plot with winter cmap and red mean line')
ax.set_xlabel('X values')
ax.set_ylabel('Y values')

plt.show()

```

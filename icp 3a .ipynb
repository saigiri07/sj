```python
import numpy as np

vec = np.random.randint(1, 21, 15)
print("Vector:\n", vec)

mat = vec.reshape(3, 5)
print("\nReshaped to 3x5:\n", mat)

print("\nShape:", mat.shape)

for i in range(mat.shape[0]):
    mat[i, np.argmax(mat[i])] = 0
print("\nAfter replacing row max with 0:\n", mat)
```

```python
# Create 4x3 array with 4-byte integers
array_4x3 = np.zeros((4, 3), dtype=np.int32)

print("Array:\n", array_4x3)
print("Shape:", array_4x3.shape)
print("Type:", type(array_4x3))
print("Data type:", array_4x3.dtype)
```

```python

matrix = np.array([[3, -2], [1, 0]])

eigenvalues, eigenvectors = np.linalg.eig(matrix)

print("Eigenvalues:", eigenvalues)
print("Right Eigenvectors:\n", eigenvectors)
```

```python

arr = np.array([[0, 1, 2], [3, 4, 5]])

diag_sum = np.trace(arr)

print("Sum of diagonal elements:", diag_sum)
```

```python

original = np.array([[1, 2], [3, 4], [5, 6]])

reshaped = original.reshape(2, 3)

print("Original 3x2:\n", original)
print("Reshaped to 2x3:\n", reshaped)
```

```python
import re

# Prompt for input
input_text = input("Enter a sentence: ")

# Replace whole word 'python' with 'pythons'
output_text = re.sub(r'\bpython\b', 'pythons', input_text)

# Output result
print("Modified sentence:")
print(output_text)
```

```python

score = float(input("Enter your class score (0–100): "))

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'

print("Your grade is:", grade)
```

```python
# Initial sets and list
IT_companies = {'Facebook', 'Google', 'Microsoft', 'Apple', 'IBM', 'Oracle', 'Amazon'}
A = {19, 22, 24, 20, 25, 26}
B = {19, 22, 20, 25, 26, 24, 28, 27}
age = [22, 19, 24, 25, 26, 24, 25, 24]

# 1. Length of IT_companies
print("Length of IT_companies:", len(IT_companies))

# 2. Add 'Twitter' to IT_companies
IT_companies.add('Twitter')
print("After adding Twitter:", IT_companies)

# 3. Insert multiple IT companies at once
IT_companies.update(['Tesla', 'Intel', 'Samsung'])
print("After adding multiple companies:", IT_companies)

# 4. Remove one company
IT_companies.remove('Oracle')
print("After removing Oracle:", IT_companies)

print("\nremove vs discard:")
temp_set = {'A', 'B'}
temp_set.discard('Z')
print("discard did not raise an error")

# 6. Join A and B (union)
print("Union of A and B:", A.union(B))

# 7. Intersection of A and B
print("Intersection of A and B:", A.intersection(B))

# 8. Is A subset of B?
print("A is subset of B:", A.issubset(B))

# 9. Are A and B disjoint
print("A and B are disjoint:", A.isdisjoint(B))

# 10. Join A with B and B with A
print("A union B:", A | B)
print("B union A:", B | A)

# 11. Symmetric difference (elements in A or B but not both)
print("Symmetric difference:", A.symmetric_difference(B))

# 12. Delete the sets completely
del A
del B

# 13. Convert age list to a set and compare lengths
age_set = set(age)
print("Length of age list:", len(age))
print("Length of age set:", len(age_set))

```

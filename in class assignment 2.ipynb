```python
class Employee:
    emp_count = 0  # class variable to count number of employees

    def __init__(self, name, family, salary, department):
        self.name = name
        self.family = family
        self.salary = salary
        self.department = department
        Employee.emp_count += 1
    @staticmethod
    def average_salary(employees):
        total_salary = sum(emp.salary for emp in employees)
        return total_salary / len(employees) if employees else 0


class FulltimeEmployee(Employee):
    pass


# Create instances
e1 = Employee("arjun", "j", 14000, "hod")
e2 = FulltimeEmployee("sai", "giri", 20000, "principal")
e3 = FulltimeEmployee("aj", "t", 6000, "a.prof")

# Use the average_salary function
all_employees = [e1, e2, e3]
print("Average Salary:", Employee.average_salary(all_employees))
print("Total Employees:", Employee.emp_count)
```
